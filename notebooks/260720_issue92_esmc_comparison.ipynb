## Introduction
In this project we set up the hierarchical UMAPs for [petadex.net](https://petadex.net). The process involves mapping the hierarchy of 60% and 30% clusters, embedding them, UMAPing them, and storing them in a way for efficient access of any UMAP

## Objectives
We have 4 goals for this project:
1. Embed all 30% and 60% cluster centroids (can be extended to 90%s as well in the future)
2. UMAP each cluster independently to allow users on the website to "click through" the UMAP
3. Build a hierarchy of cluster centroids that is easier to use than the current system via the PostGres `petadex_clustering` table (slow) and the catalytic_orfs file (large)
4. Compare the UMAP embeddings of _BioHub_'s ESMC and the `igem-toronto` finetuned version

## Methods
- **Reducer**: `umap-learn`, `n_neighbors=15`, `min_dist=0.1`, `metric=euclidean`, `random_state=42`.
