## Introduction

Jessica Shen already made a plot of raw amounts of PETadex sequences and where they came from, which is this:

<img src="../resources/260818_issue107_sample_density_plot/outputs/petadex_catalytic_orf_density_moll.svg" style="background-color: white;" />

We are concerned that this plot just shows us where sampling is happening, and is not PETase specific. We want to remake this plot, normalized by the number of Logan Biosamples to see if we can identify areas with abnormally high amounts of PETases for the amount of sampling being done.

## Objectives

1. Pull all PETadex orfs with their coords from `petadex.sra_metadata` and `petadex.logan_catalytic_orfs`, as well as all Logan biosamples from `logan.biosample_geographical_location`
2. Map all longitude-latitude coordinates to x-y coordinates using Mollweide projection
3. Split Earth into a grid, and count all orfs and all biosamples per grid box
4. Plot the amount of PETadex samples per grid box, normalized by amount
